### Aim
Demonstrate the Data science life cycle and perform Univariate , bivariate and multivariate analysis on a real-world dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("employee.csv")
```

### Select Numeric Columns

```python
numeric_cols = df.select_dtypes(include='number').columns
```

### 1.Univariate

```python
df[numeric_cols[0]].describe()
```

### 2.Bivariate

```python
correlation = df[numeric_cols[0]].corr(df[numeric_cols[1]])

print("Correlation between", numeric_cols[0], "and", numeric_cols[1], "=", correlation)
```

### 3.Multivariate

```python
variables = numeric_cols[:3]
print("Selected variables:")
print(variables)
correlation_matrix = df[variables].corr()
print(correlation_matrix)
```
