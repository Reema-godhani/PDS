# Practical 1
### Aim
To write Python programs to demonstrate variables, basic data types and different types of operators.

### 1. Variables and Data Types

```python
name = "reema"
roll_no = 54
age = 19
marks = 85.5
student = True

print(name)
print(roll_no)
print(age)
print(marks)
print(student)

print(type(name))
print(type(roll_no))
print(type(age))
print(type(marks))
print(type(student))
```

### 2. Arithmetic Operators

```python
a = 20
b = 6

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

### 3. Assignment Operators

```python
a = 10
print(a)

a += 5
print(a)

a -= 2
print(a)

a *= 2
print(a)

a /= 2
print(a)

a //= 2
print(a)

a %= 3
print(a)
```

### 4. Comparison Operators

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

### 5. Logical Operators

```python
a = 10
b = 20

print(a > 5 and b > 15)
print(a > 15 or b > 15)
print(not(a > 15))
```

### 6. Membership Operators

```python
name = "Ashish"
numbers = [10, 20, 30, 40]

print("A" in name)
print("z" not in name)
print(20 in numbers)
print(50 not in numbers)
```

### 7. Identity Operators

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a is not c)
```

### 8. Bitwise Operators

```python
a = 10
b = 6

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)
```
